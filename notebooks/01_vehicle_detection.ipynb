# 🚗 Vehicle Detection Demo

This notebook demonstrates the vehicle detection system using YOLOv8.

## Setup

```python
import sys
sys.path.insert(0, '..')  # Add parent directory

import cv2
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

from src.detection import VehicleDetector
from src.utils.image_utils import draw_bbox
```

## Initialize Detector

```python
# Create detector
detector = VehicleDetector()

# Get info
info = detector.get_info()
print(f"Model: {info['model_type']}")
print(f"Device: {info['device']}")
print(f"Confidence threshold: {info['confidence_threshold']}")
```

## Load Test Image

```python
# Load your image here
image_path = "../data/samples/test_image.jpg"

# Create dummy image if file doesn't exist
if not Path(image_path).exists():
    print("⚠️  Sample image not found. Using dummy image.")
    image = np.random.randint(0, 255, (640, 640, 3), dtype=np.uint8)
else:
    image = cv2.imread(image_path)
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)  # For matplotlib

# Display
plt.figure(figsize=(12, 8))
plt.imshow(image)
plt.title("Original Image")
plt.axis('off')
plt.show()
```

## Detect Vehicles

```python
# Convert back to BGR for detector
image_bgr = cv2.cvtColor(image, cv2.COLOR_RGB2BGR)

# Detect
detections = detector.detect_vehicles(image_bgr)

print(f"✓ Detected {len(detections)} vehicles")

# Show detections
for i, det in enumerate(detections, 1):
    print(f"{i}. {det['class_name']} - Confidence: {det['confidence']:.2%}")
```

## Visualize Results

```python
# Draw bounding boxes
result_image = image_bgr.copy()

for det in detections:
    label = f"{det['class_name']} {det['confidence']:.2f}"
    result_image = draw_bbox(
        result_image,
        det['bbox'],
        label=label,
        color=(0, 255, 0)
    )

# Convert to RGB for display
result_image = cv2.cvtColor(result_image, cv2.COLOR_BGR2RGB)

# Display
plt.figure(figsize=(12, 8))
plt.imshow(result_image)
plt.title(f"Detection Results ({len(detections)} vehicles)")
plt.axis('off')
plt.show()
```

## Analysis

```python
# Count by vehicle type
from collections import Counter

vehicle_types = [det['class_name'] for det in detections]
type_counts = Counter(vehicle_types)

print("Vehicle Type Distribution:")
for vtype, count in type_counts.items():
    print(f"  {vtype}: {count}")

# Average confidence
if detections:
    avg_conf = np.mean([det['confidence'] for det in detections])
    print(f"\nAverage Confidence: {avg_conf:.2%}")
```
